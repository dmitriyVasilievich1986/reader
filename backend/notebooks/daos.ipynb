## Import libraries

```python
from dotenv import load_dotenv
from pathlib import Path
```

```python
from reader.config import AppConfig
from reader.services.database import AsyncDatabaseClient
```

```python
from reader.services.daos import BookDAO, AuthorDAO, CategoryDAO, PageDAO
```

## Initialize constants

```python
load_dotenv(Path("../.env"))
config = AppConfig()
```

```python
config.services.database.host = str(config.info.paths_info.root / "reader.sqlite3")
db_client = AsyncDatabaseClient(config)
```
